# 05c Title card and legends

Phase 5c check: the title card, `scheme-linear` with its gradient legend, the bar legend, the height key, and cleaner 3D colours. Record pass or fail in the table at the end.

```python
import warnings
import webbrowser
from pathlib import Path

import geopandas as gpd

import kharita as kh

warnings.filterwarnings("ignore", module="lonboard")
states = gpd.read_file("../data/processed/sudan_states.gpkg")
buildings = gpd.read_file("../data/processed/khartoum_buildings.gpkg")
SOURCE = "U.S. Census Bureau via HDX, CC BY 4.0. 2008 census, retabulated."
OVERTURE = "Overture Maps Foundation, OpenStreetMap contributors (ODbL)."
```

```python
# A. Title card top left, legend bottom right
kh.map(
    states,
    "preset-choropleth fill-[density] ramp-ylorrd tooltip-[key,density]",
    legend={"title": "People per km2"},
    title="Sudan population density",
    subtitle="People per square kilometre by state",
    source=SOURCE,
)
```

```python
# B. Title and legend in the same corner stack; dark theme styles both
kh.map(
    states,
    "theme-midnight fill-[density] scheme-q5 legend-tl halo-cyan-300",
    legend={"title": "People per km2"},
    title="Where Sudan lives",
    source=SOURCE,
)
```

```python
# C. No classes: scheme-linear with a gradient legend
kh.map(
    states,
    "theme-paper fill-[density] scheme-linear ramp-viridis stroke-white",
    legend={"title": "People per km2", "fmt": "{:,.0f}"},
    title="Density without classes",
)
```

```python
# D. Bar legend: the classes as one strip with their edges underneath
kh.map(
    states,
    "preset-choropleth fill-[pop] ramp-blues scheme-q5",
    legend={"title": "Population", "style": "bar", "fmt": "{:,.0f}"},
    title="Population by state",
)
```

```python
# E. 3D states: ylorrd now skips its near-white end, edges follow the fill, height key in the legend
kh.map(
    states,
    "theme-paper fill-[density] ramp-ylorrd scheme-q5 extrude-[pop]",
    legend={"title": "People per km2"},
    title="Population as height, density as colour",
)
```

```python
# F. Khartoum buildings again: no white speckle, height key shows real metres
tallest = buildings["height_m"].max()
kh.map(
    buildings,
    f"theme-midnight fill-[height_m] scheme-[7,12,20,40] extrude-[height_m] extrude-max-{tallest:g} buildings-none",
    legend={"title": "Height, m", "note": "Unmapped heights assumed 6 m."},
    title="Central Khartoum",
    subtitle="Building heights",
    source=OVERTURE,
)
```

```python
# G. Export: the title card and legend travel into the HTML file, and the page takes the title
out = Path("../spike_out")
out.mkdir(exist_ok=True)
final = kh.map(
    states,
    "theme-midnight fill-[density] scheme-linear tooltip-[key,density]",
    legend={"title": "People per km2", "fmt": "{:,.0f}"},
    title="Sudan population density",
    subtitle="People per square kilometre",
    source=SOURCE,
)
final.to_html(out / "sudan_titled.html")
webbrowser.open((out / "sudan_titled.html").resolve().as_uri())
```
