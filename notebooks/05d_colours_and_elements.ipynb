# 05d Colour system and map elements

Phase 5d check: about 160 named ramps, ramps from Tailwind shades, hex colours, fixed category colours, standard palettes, your own palettes and themes, map element colours and stepped relief. Record pass or fail in the table at the end.

```python
import warnings
import webbrowser
from pathlib import Path

import geopandas as gpd
from IPython.display import display
from lonboard import Map

import kharita as kh
from kharita import basemaps as kb

warnings.filterwarnings("ignore", module="lonboard")
states = gpd.read_file("../data/processed/sudan_states.gpkg")
roads = gpd.read_file("../data/processed/khartoum_roads.gpkg")
buildings = gpd.read_file("../data/processed/khartoum_buildings.gpkg")
major = roads[roads["rank"] >= 4]
OVERTURE = "Overture Maps Foundation, OpenStreetMap contributors (ODbL)."
```

```python
# A. Every Tailwind shade, then every ramp, as swatches
display(kh.colors())
kh.ramps()
```

```python
# B. Narrow the list, and check a ramp
display(kh.ramps(kind="diverging", source="crameri"))
display(kh.ramps(search="sun"))
kh.check_ramp("batlow"), kh.check_ramp("set2")
```

```python
# C. Any hex colour, in brackets, with opacity
kh.map(states, "fill-[#0f766e]/70 stroke-[#f8fafc] stroke-1.5 basemap-paper", title="Hex colours")
```

```python
# D. Four ramps from four collections on the same data
for ramp in ("batlow", "sunset", "oslo", "lajolla"):
    display(kh.map(states, f"theme-paper fill-[density] ramp-{ramp} scheme-q5", legend={"title": "People per km2", "style": "bar"}, title=f"ramp-{ramp}", height=320))
```

```python
# E. Ramps from Tailwind shades: one family, two families around the mean, and your own stops
mean = round(states["density"].mean(), 1)
display(kh.map(states, "theme-paper fill-[density] ramp-teal scheme-q5", title="ramp-teal", height=320))
display(kh.map(states, f"theme-paper fill-[density] ramp-teal-rose scheme-div-[{mean}]", title=f"ramp-teal-rose around the mean ({mean})", height=320))
display(kh.map(states, "theme-midnight fill-[density] ramp-[sky-200,indigo-500,fuchsia-600] scheme-linear", title="Your own stops, unclassed", height=320))
```

```python
# F. Fixed colours for chosen building types; the rest from the ramp, in your order
kh.map(
    buildings,
    "theme-mono fill-[kind] ramp-pastel stroke-0.3 stroke-white",
    colors={"commercial": "rose-500", "religious": "emerald-600", "education": "sky-600"},
    order=["commercial", "religious", "education"],
    legend={"title": "Building type", "note": OVERTURE},
    title="Fixed category colours",
)
```

```python
# G. A standard palette: IPC phases keep their official colours and labels
# The phases below are made up to show the palette; they are not real IPC data
illustrative = states.assign(phase=[(i % 5) + 1 for i in range(len(states))])
kh.map(
    illustrative,
    "preset-choropleth fill-[phase] ramp-ipc",
    legend={"title": "IPC phase", "note": "Illustrative values, not real IPC data."},
    title="ramp-ipc",
)
```

```python
# H. Your own palette and your own theme
kh.palette("nile", ["#0c4a6e", "teal-400", "amber-200", "orange-700"])
kh.theme("desert", base="paper", land="[#f3e5c8]", water="sky-300", roads="amber-200", borders="stone-600", ramp="nile")
kh.map(states, "theme-desert fill-[density] scheme-q5 stroke-white", legend={"title": "People per km2"}, title="theme-desert with ramp-nile")
```

```python
# I. Recolour the basemap element by element: Khartoum on muted
kh.map(
    major,
    "stroke-2 stroke-rose-500 basemap-muted land-[#f5f1ea] water-sky-300 roads-amber-200 "
    "streets-stone-200 buildings-stone-300 parks-emerald-100 labels-slate-700 legend-none",
    title="Central Khartoum",
    subtitle="Every basemap element recoloured",
    source=OVERTURE,
)
```

```python
# J. Stepped relief: flat bands by height, at Jebel Marra and across Sudan
JEBEL_MARRA = {"longitude": 24.25, "latitude": 12.95, "zoom": 8}
display(Map(kh.layer(states, "fill-transparent stroke-1 stroke-slate-700"), basemap=kb.basemap("paper", relief="steps"), view_state=JEBEL_MARRA, height=360))
kh.map(states, "preset-outline stroke-slate-700 basemap-paper relief-steps halo-white", title="relief-steps")
```

```python
# K. Bar legend fix from 05c test D: long numbers no longer overflow
kh.map(
    states,
    "preset-choropleth fill-[pop] ramp-blues scheme-q5",
    legend={"title": "Population", "style": "bar", "fmt": "{:,.0f}"},
    title="Population by state",
)
```

```python
# L. The ground warning: palest grey would vanish into the mono ground
warnings.simplefilter("always", kh.KharitaWarning)
kh.map(states, "fill-[density] basemap-mono ramp-greys scheme-q5", title="Expect a warning above this map")
```

```python
# M. Save the ramp sheet as a page you can keep open while styling
out = Path("../spike_out")
out.mkdir(exist_ok=True)
kh.ramps().to_html(out / "kharita_ramps.html")
webbrowser.open((out / "kharita_ramps.html").resolve().as_uri())
```
